```python
import pandas as pd
from utils import *
```

```python
df_ = pd.read_csv('predicted_data_3.csv')
```

```python
df = df_.copy()
df.head()
```

```python
agg_df = (
    df
    .groupby(['year', 'month', 'blid'])
    .agg(
        mean_actual=('mietekalt', 'mean'),
        mean_predicted=('predicted_mietekalt', 'mean'),
        mean_difference=('difference', 'mean'),
        median_difference=('difference', 'median'),
        std_difference=('difference', 'std'),
        count=('difference', 'count')
    )
    .reset_index()
).round(2)
```
